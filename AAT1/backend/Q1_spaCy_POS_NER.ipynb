**Aim:** Use spaCy to perform POS tagging and named entity recognition (NER) on a paragraph and display the words, their POS tags and the named entities.

```python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")
print("spaCy", spacy.__version__, "| model", nlp.meta["name"], nlp.meta["version"])
```

```python
from pathlib import Path
DATA = Path("../data")
paragraph = (DATA / "paragraph.txt").read_text(encoding="utf-8").strip()
print("Loaded", (DATA / "paragraph.txt").resolve())

doc = nlp(paragraph)
```

```python
# Words and their POS tags
print(f"{'Word':<15}{'POS':<8}{'Tag':<6}{'Lemma':<15}Meaning")
print("-" * 62)
for token in doc:
    print(f"{token.text:<15}{token.pos_:<8}{token.tag_:<6}{token.lemma_:<15}{spacy.explain(token.pos_)}")
```

```python
# Named entities
print(f"{'Entity':<22}{'Label':<10}Meaning")
print("-" * 60)
for ent in doc.ents:
    print(f"{ent.text:<22}{ent.label_:<10}{spacy.explain(ent.label_)}")
```

```python
# Count of each POS tag in the paragraph
from collections import Counter
print(Counter(token.pos_ for token in doc).most_common())
```

```python
# Highlighted entity view (renders inside Jupyter)
displacy.render(doc, style="ent", jupyter=True)
```

```python
# Try another paragraph stored in data/extra_paragraph.txt
my_text = (DATA / "extra_paragraph.txt").read_text(encoding="utf-8").strip()
print(my_text)
for ent in nlp(my_text).ents:
    print(ent.text, "->", ent.label_)
```
